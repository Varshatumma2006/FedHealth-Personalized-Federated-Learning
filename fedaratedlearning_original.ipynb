```python
import pandas as pd
import numpy as np

# Dataset path
file_path = "breast-cancer.data"

# Column names
columns = [
    "Class",
    "Age",
    "Menopause",
    "Tumor-size",
    "Inv-nodes",
    "Node-caps",
    "Deg-malig",
    "Breast",
    "Breast-quad",
    "Irradiat"
]

# Load dataset
df = pd.read_csv(file_path, header=None, names=columns)

# Display basic information
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nClass Distribution:")
print(df["Class"].value_counts())
```

```python
# Remove extra spaces from categorical values
for column in df.columns:
    if df[column].dtype == "object":
        df[column] = df[column].str.strip()

# Replace '?' with NaN
df = df.replace("?", np.nan)

# Check missing values again
print("Missing values after cleaning:")
print(df.isnull().sum())

# Remove rows containing missing values
df = df.dropna().reset_index(drop=True)

print("\nDataset after cleaning:")
print(df.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

df["Class"] = label_encoder.fit_transform(df["Class"])

print(df["Class"].value_counts())
print("\nClasses:")
print(label_encoder.classes_)
```

```python
# STEP 4: ENCODE CATEGORICAL FEATURES

from sklearn.preprocessing import LabelEncoder

df_encoded = df.copy()

categorical_columns = df_encoded.select_dtypes(include=["object"]).columns

encoders = {}

for column in categorical_columns:
    encoder = LabelEncoder()
    df_encoded[column] = encoder.fit_transform(df_encoded[column])
    encoders[column] = encoder

print("Encoded Dataset:")
print(df_encoded.head())

print("\nData Types:")
print(df_encoded.dtypes)

print("\nEncoded Class Distribution:")
print(df_encoded["Class"].value_counts())

print("\nEncoded Dataset Shape:")
print(df_encoded.shape)
```

```python
# ============================================
# STEP 5: SEPARATE FEATURES AND TARGET
# ============================================

X = df_encoded.drop("Class", axis=1)
y = df_encoded["Class"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# ============================================
# STEP 6: FEATURE SCALING
# ============================================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Original X shape:", X.shape)
print("Scaled X shape:", X_scaled.shape)

print("\nFirst 5 scaled rows:")
print(X_scaled[:5])
```

```python
# ============================================
# STEP 7: TRAIN / TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("\nTraining target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
# ============================================
# STEP 8: CREATE NON-IID HEALTHCARE CLIENTS
# ============================================

import numpy as np
import pandas as pd

# Number of federated clients
NUM_CLIENTS = 5

# Non-IID concentration parameter
# Smaller alpha = stronger Non-IID distribution
alpha = 0.5

# Combine training features and labels
train_X = np.array(X_train)
train_y = np.array(y_train)

# Store client datasets
client_data = []

# Random generator
rng = np.random.default_rng(42)

# Create client indices
client_indices = [[] for _ in range(NUM_CLIENTS)]

# Process each class separately
for class_label in np.unique(train_y):

    # Get indices belonging to this class
    class_indices = np.where(train_y == class_label)[0]

    # Shuffle indices
    rng.shuffle(class_indices)

    # Generate Dirichlet proportions
    proportions = rng.dirichlet(
        np.repeat(alpha, NUM_CLIENTS)
    )

    # Convert proportions into sample counts
    counts = (proportions * len(class_indices)).astype(int)

    # Correct rounding difference
    difference = len(class_indices) - counts.sum()

    for i in range(difference):
        counts[i % NUM_CLIENTS] += 1

    # Assign samples to clients
    start = 0

    for client_id in range(NUM_CLIENTS):

        end = start + counts[client_id]

        client_indices[client_id].extend(
            class_indices[start:end]
        )

        start = end


# Create client datasets
for client_id in range(NUM_CLIENTS):

    indices = np.array(client_indices[client_id])

    # Shuffle client samples
    rng.shuffle(indices)

    client_X = train_X[indices]
    client_y = train_y[indices]

    client_data.append(
        {
            "X": client_X,
            "y": client_y
        }
    )


# ============================================
# DISPLAY CLIENT DISTRIBUTION
# ============================================

print("=" * 60)
print("NON-IID CLIENT DISTRIBUTION")
print("=" * 60)

for i, client in enumerate(client_data):

    X_client = client["X"]
    y_client = client["y"]

    class_0 = np.sum(y_client == 0)
    class_1 = np.sum(y_client == 1)

    print(f"\nClient {i + 1}")
    print("-" * 30)
    print("Total samples :", len(y_client))
    print("Class 0       :", class_0)
    print("Class 1       :", class_1)

    if len(y_client) > 0:
        print(
            "Class 0 ratio :",
            round(class_0 / len(y_client), 3)
        )

        print(
            "Class 1 ratio :",
            round(class_1 / len(y_client), 3)
        )
```

```python
# ============================================
# STEP 9: BUILD THE BASE NEURAL NETWORK
# ============================================

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input

def create_model():

    model = Sequential([
        Input(shape=(9,)),

        Dense(32, activation="relu"),

        Dense(16, activation="relu"),

        Dense(1, activation="sigmoid")
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


# Create a test model
model = create_model()

model.summary()
```

```python
# ============================================
# STEP 10: CENTRALIZED BASELINE
# ============================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Create fresh model
central_model = create_model()

# Train centralized model
history = central_model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=16,
    validation_split=0.15,
    verbose=0
)

# Predictions
y_pred_prob = central_model.predict(
    X_test,
    verbose=0
)

y_pred = (y_pred_prob >= 0.5).astype(int).flatten()

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("=" * 50)
print("CENTRALIZED BASELINE RESULTS")
print("=" * 50)

print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1 * 100:.2f}%")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "No Recurrence",
            "Recurrence"
        ],
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# ============================================
# STEP 11: LOCAL CLIENT TRAINING
# ============================================

def train_local_model(client_X, client_y, epochs=20):

    # Create a fresh model
    local_model = create_model()

    # Train only on this client's data
    local_model.fit(
        client_X,
        client_y,
        epochs=epochs,
        batch_size=8,
        verbose=0
    )

    return local_model


# Train one local model for each client
local_models = []

print("=" * 60)
print("LOCAL CLIENT TRAINING")
print("=" * 60)

for client_id, client in enumerate(client_data):

    X_client = client["X"]
    y_client = client["y"]

    print(
        f"\nClient {client_id + 1}: "
        f"{len(y_client)} samples"
    )

    local_model = train_local_model(
        X_client,
        y_client,
        epochs=20
    )

    local_models.append(local_model)

    print("Local training completed.")
```

```python
# ============================================
# STEP 12A: FEDAVG AGGREGATION
# ============================================

def fedavg(local_models, client_data):

    # Total number of training samples
    total_samples = sum(
        len(client["y"])
        for client in client_data
    )

    # Get weights from the first model
    global_weights = [
        np.zeros_like(weight)
        for weight in local_models[0].get_weights()
    ]

    # Weighted average
    for model, client in zip(local_models, client_data):

        client_samples = len(client["y"])

        # Client contribution
        weight_factor = (
            client_samples / total_samples
        )

        model_weights = model.get_weights()

        for i in range(len(global_weights)):

            global_weights[i] += (
                weight_factor * model_weights[i]
            )

    return global_weights
    # ============================================
# STEP 12B: CREATE GLOBAL MODEL
# ============================================

# Create global model
global_model = create_model()

# Aggregate client models
global_weights = fedavg(
    local_models,
    client_data
)

# Set aggregated weights
global_model.set_weights(global_weights)

print("=" * 60)
print("FEDAVG GLOBAL MODEL CREATED")
print("=" * 60)

print("Number of clients:", len(local_models))
print(
    "Total training samples:",
    sum(len(client["y"]) for client in client_data)
)

print("Global model parameters:",
      global_model.count_params())
# ============================================
# STEP 12C: EVALUATE FEDAVG
# ============================================

fedavg_prob = global_model.predict(
    X_test,
    verbose=0
)

fedavg_pred = (
    fedavg_prob >= 0.5
).astype(int).flatten()

# Calculate metrics
fedavg_accuracy = accuracy_score(
    y_test,
    fedavg_pred
)

fedavg_precision = precision_score(
    y_test,
    fedavg_pred,
    zero_division=0
)

fedavg_recall = recall_score(
    y_test,
    fedavg_pred,
    zero_division=0
)

fedavg_f1 = f1_score(
    y_test,
    fedavg_pred,
    zero_division=0
)

print("=" * 60)
print("FEDAVG RESULTS")
print("=" * 60)

print(f"Accuracy  : {fedavg_accuracy * 100:.2f}%")
print(f"Precision : {fedavg_precision * 100:.2f}%")
print(f"Recall    : {fedavg_recall * 100:.2f}%")
print(f"F1 Score  : {fedavg_f1 * 100:.2f}%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, fedavg_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        fedavg_pred,
        target_names=[
            "No Recurrence",
            "Recurrence"
        ],
        zero_division=0
    )
)
```

```python
# ============================================
# STEP 13: MULTI-ROUND FEDAVG
# ============================================

import numpy as np
import tensorflow as tf

NUM_ROUNDS = 20
LOCAL_EPOCHS = 10
BATCH_SIZE = 8

# Create initial global model
fedavg_model = create_model()

# Store results
fedavg_history = {
    "round": [],
    "accuracy": [],
    "precision": [],
    "recall": [],
    "f1": []
}

print("=" * 70)
print("MULTI-ROUND FEDAVG TRAINING")
print("=" * 70)

for round_num in range(1, NUM_ROUNDS + 1):

    local_weights = []
    client_sizes = []

    # ----------------------------------------
    # Local training
    # ----------------------------------------

    for client_id, client in enumerate(client_data):

        X_client = client["X"]
        y_client = client["y"]

        # Create local model
        local_model = create_model()

        # Start from current global model
        local_model.set_weights(
            fedavg_model.get_weights()
        )

        # Train locally
        local_model.fit(
            X_client,
            y_client,
            epochs=LOCAL_EPOCHS,
            batch_size=BATCH_SIZE,
            verbose=0
        )

        # Store local weights
        local_weights.append(
            local_model.get_weights()
        )

        client_sizes.append(
            len(y_client)
        )

    # ----------------------------------------
    # FedAvg aggregation
    # ----------------------------------------

    total_samples = sum(client_sizes)

    new_global_weights = []

    for layer_index in range(
        len(local_weights[0])
    ):

        weighted_layer = np.zeros_like(
            local_weights[0][layer_index]
        )

        for client_index in range(
            len(local_weights)
        ):

            weight = (
                client_sizes[client_index]
                / total_samples
            )

            weighted_layer += (
                weight *
                local_weights[client_index][layer_index]
            )

        new_global_weights.append(
            weighted_layer
        )

    # Update global model
    fedavg_model.set_weights(
        new_global_weights
    )

    # ----------------------------------------
    # Evaluate global model
    # ----------------------------------------

    predictions = fedavg_model.predict(
        X_test,
        verbose=0
    )

    predictions = (
        predictions >= 0.5
    ).astype(int).flatten()

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    # Store results
    fedavg_history["round"].append(
        round_num
    )

    fedavg_history["accuracy"].append(
        accuracy * 100
    )

    fedavg_history["precision"].append(
        precision * 100
    )

    fedavg_history["recall"].append(
        recall * 100
    )

    fedavg_history["f1"].append(
        f1 * 100
    )

    print(
        f"Round {round_num:02d} | "
        f"Accuracy: {accuracy * 100:6.2f}% | "
        f"Precision: {precision * 100:6.2f}% | "
        f"Recall: {recall * 100:6.2f}% | "
        f"F1: {f1 * 100:6.2f}%"
    )
```

```python
# ============================================
# FEDAVG FINAL SUMMARY
# ============================================

fedavg_results_df = pd.DataFrame(fedavg_history)

# Find best round based on F1 score
best_round = fedavg_results_df.loc[
    fedavg_results_df["f1"].idxmax()
]

print("\n" + "=" * 70)
print("FEDAVG FINAL SUMMARY")
print("=" * 70)

print(f"Best Round  : {int(best_round['round'])}")
print(f"Accuracy    : {best_round['accuracy']:.2f}%")
print(f"Precision   : {best_round['precision']:.2f}%")
print(f"Recall      : {best_round['recall']:.2f}%")
print(f"F1 Score    : {best_round['f1']:.2f}%")

print("=" * 70)
```

```python
# ============================================
# STEP 15A: LIGHTWEIGHT PERSONALIZATION
# ============================================

def create_personalized_model(global_model):

    # Create a fresh model
    model = create_model()

    # Copy global weights
    model.set_weights(
        global_model.get_weights()
    )

    # Freeze shared feature layers
    model.layers[0].trainable = False
    model.layers[1].trainable = False

    # Keep final classification layer trainable
    model.layers[2].trainable = True

    # Recompile after changing trainable status
    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.0005
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model
    # ============================================
# STEP 15B: TRAIN LIGHTWEIGHT PERSONALIZED MODELS
# ============================================

PERSONALIZATION_EPOCHS = 5

light_personalized_models = []

print("=" * 70)
print("LIGHTWEIGHT PERSONALIZED LOCAL ADAPTATION")
print("=" * 70)

for client_id, client in enumerate(client_data):

    X_client = client["X"]
    y_client = client["y"]

    # Create personalized model from FedAvg
    personalized_model = create_personalized_model(
        fedavg_model
    )

    # Train only personalized layer
    personalized_model.fit(
        X_client,
        y_client,
        epochs=PERSONALIZATION_EPOCHS,
        batch_size=8,
        verbose=0
    )

    light_personalized_models.append(
        personalized_model
    )

    print(
        f"Client {client_id + 1}: "
        f"{len(y_client)} samples | "
        f"Adaptation completed"
    )
```

```python
# ============================================
# STEP 16A: CALCULATE CLASS WEIGHTS
# ============================================

from sklearn.utils.class_weight import compute_class_weight

classes = np.unique(y_train)

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = {
    int(classes[i]): float(class_weights_array[i])
    for i in range(len(classes))
}

print("=" * 60)
print("CLASS WEIGHTS")
print("=" * 60)

print("Class 0 weight:", class_weights[0])
print("Class 1 weight:", class_weights[1])
# ============================================
# STEP 16B: WEIGHTED LOCAL TRAINING
# ============================================

def train_weighted_local_model(
    client_X,
    client_y,
    initial_weights,
    epochs=10
):

    model = create_model()

    # Start from global model
    model.set_weights(initial_weights)

    # Train using class weights
    model.fit(
        client_X,
        client_y,
        epochs=epochs,
        batch_size=8,
        class_weight=class_weights,
        verbose=0
    )

    return model
    # ============================================
# STEP 16C: CLASS-WEIGHTED FEDAVG
# ============================================

NUM_ROUNDS_WEIGHTED = 20
LOCAL_EPOCHS_WEIGHTED = 10

weighted_fedavg_model = create_model()

weighted_history = {
    "round": [],
    "accuracy": [],
    "precision": [],
    "recall": [],
    "f1": []
}

print("=" * 70)
print("CLASS-WEIGHTED MULTI-ROUND FEDAVG")
print("=" * 70)

for round_num in range(1, NUM_ROUNDS_WEIGHTED + 1):

    local_weights = []
    client_sizes = []

    for client in client_data:

        X_client = client["X"]
        y_client = client["y"]

        local_model = train_weighted_local_model(
            X_client,
            y_client,
            weighted_fedavg_model.get_weights(),
            epochs=LOCAL_EPOCHS_WEIGHTED
        )

        local_weights.append(
            local_model.get_weights()
        )

        client_sizes.append(
            len(y_client)
        )

    # FedAvg aggregation
    total_samples = sum(client_sizes)

    new_weights = []

    for layer_index in range(len(local_weights[0])):

        averaged_layer = np.zeros_like(
            local_weights[0][layer_index]
        )

        for client_index in range(len(local_weights)):

            contribution = (
                client_sizes[client_index]
                / total_samples
            )

            averaged_layer += (
                contribution *
                local_weights[client_index][layer_index]
            )

        new_weights.append(averaged_layer)

    weighted_fedavg_model.set_weights(new_weights)

    # Evaluation
    probabilities = weighted_fedavg_model.predict(
        X_test,
        verbose=0
    )

    predictions = (
        probabilities >= 0.5
    ).astype(int).flatten()

    acc = accuracy_score(y_test, predictions)
    prec = precision_score(
        y_test, predictions, zero_division=0
    )
    rec = recall_score(
        y_test, predictions, zero_division=0
    )
    f1_value = f1_score(
        y_test, predictions, zero_division=0
    )

    weighted_history["round"].append(round_num)
    weighted_history["accuracy"].append(acc * 100)
    weighted_history["precision"].append(prec * 100)
    weighted_history["recall"].append(rec * 100)
    weighted_history["f1"].append(f1_value * 100)

    print(
        f"Round {round_num:02d} | "
        f"Accuracy: {acc * 100:6.2f}% | "
        f"Precision: {prec * 100:6.2f}% | "
        f"Recall: {rec * 100:6.2f}% | "
        f"F1: {f1_value * 100:6.2f}%"
    )
```

```python
# ============================================
# STEP 20: CORRECT NON-IID PARTITION
# ============================================

import numpy as np
import pandas as pd

NUM_CLIENTS = 5

X_train_np = np.array(X_train)
y_train_np = np.array(y_train)

rng = np.random.default_rng(42)

# Get indices for both classes
class0_indices = np.where(y_train_np == 0)[0]
class1_indices = np.where(y_train_np == 1)[0]

# Shuffle
rng.shuffle(class0_indices)
rng.shuffle(class1_indices)

# --------------------------------------------
# Desired number of samples per client
# --------------------------------------------

client_sizes = [44, 44, 44, 44, 45]

# --------------------------------------------
# Allocate Class 1 samples
# --------------------------------------------

# Total Class 1 = 65
# Deliberately different proportions
class1_counts = [6, 10, 14, 17, 18]

print("Class 1 allocation:", class1_counts)
print("Total Class 1:", sum(class1_counts))

# Calculate Class 0 counts
class0_counts = [
    client_sizes[i] - class1_counts[i]
    for i in range(NUM_CLIENTS)
]

print("Class 0 allocation:", class0_counts)
print("Total Class 0:", sum(class0_counts))

# Verify availability
assert sum(class1_counts) == len(class1_indices)
assert sum(class0_counts) == len(class0_indices)

# --------------------------------------------
# Create clients
# --------------------------------------------

client_data_controlled = []

start0 = 0
start1 = 0

for client_id in range(NUM_CLIENTS):

    n0 = class0_counts[client_id]
    n1 = class1_counts[client_id]

    indices0 = class0_indices[
        start0:start0 + n0
    ]

    indices1 = class1_indices[
        start1:start1 + n1
    ]

    start0 += n0
    start1 += n1

    # Combine both classes
    indices = np.concatenate([
        indices0,
        indices1
    ])

    # Shuffle client data
    rng.shuffle(indices)

    client_X = X_train_np[indices]
    client_y = y_train_np[indices]

    client_data_controlled.append({
        "X": client_X,
        "y": client_y
    })


# ============================================
# CHECK DISTRIBUTION
# ============================================

print("\n" + "=" * 70)
print("CORRECTED NON-IID CLIENT DISTRIBUTION")
print("=" * 70)

total_samples = 0
total_class0 = 0
total_class1 = 0

for i, client in enumerate(
    client_data_controlled
):

    y_client = client["y"]

    class0 = np.sum(y_client == 0)
    class1 = np.sum(y_client == 1)
    total = len(y_client)

    total_samples += total
    total_class0 += class0
    total_class1 += class1

    print(f"\nClient {i + 1}")
    print("-" * 30)

    print("Total samples :", total)
    print("Class 0       :", class0)
    print("Class 1       :", class1)

    print(
        "Class 0 ratio :",
        f"{class0 / total:.3f}"
    )

    print(
        "Class 1 ratio :",
        f"{class1 / total:.3f}"
    )


# ============================================
# FINAL VERIFICATION
# ============================================

print("\n" + "=" * 70)
print("FINAL PARTITION CHECK")
print("=" * 70)

print("Total samples :", total_samples)
print("Total Class 0 :", total_class0)
print("Total Class 1 :", total_class1)

print("\nOriginal training samples :", len(y_train))
print("Original Class 0          :", np.sum(y_train_np == 0))
print("Original Class 1          :", np.sum(y_train_np == 1))

assert total_samples == len(y_train)
assert total_class0 == np.sum(y_train_np == 0)
assert total_class1 == np.sum(y_train_np == 1)

print("\n✓ All 221 training samples accounted for.")
print("✓ No samples lost.")
print("✓ No samples duplicated.")
print("✓ Every client contains both classes.")
print("✓ Non-IID distribution created.")
```

```python
# ============================================
# STEP 22A: PERSONALIZED ADAPTATION
# ============================================

from tensorflow.keras.callbacks import EarlyStopping

def create_personalized_model_v2(global_model):

    model = create_model()

    # Start from the best FedAvg model
    model.set_weights(
        global_model.get_weights()
    )

    # Freeze first shared layer
    model.layers[0].trainable = False

    # Allow local adaptation
    model.layers[1].trainable = True
    model.layers[2].trainable = True

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.0001
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model
    # ============================================
# STEP 22B: TRAIN PERSONALIZED CLIENT MODELS
# ============================================

personalized_v2_models = []

print("=" * 70)
print("PERSONALIZED LOCAL ADAPTATION - V2")
print("=" * 70)

for client_id, client in enumerate(
    client_data_controlled
):

    X_client = client["X"]
    y_client = client["y"]

    model = create_personalized_model_v2(
        best_controlled_fedavg
    )

    early_stopping = EarlyStopping(
        monitor="loss",
        patience=2,
        restore_best_weights=True
    )

    model.fit(
        X_client,
        y_client,
        epochs=20,
        batch_size=8,
        verbose=0,
        callbacks=[early_stopping]
    )

    personalized_v2_models.append(model)

    print(
        f"Client {client_id + 1} | "
        f"Samples: {len(y_client)} | "
        f"Personalization completed"
    )
    # ============================================
# STEP 22C: EVALUATE PERSONALIZED V2
# ============================================

personalized_v2_results = []

print("\n" + "=" * 70)
print("PERSONALIZED V2 RESULTS")
print("=" * 70)

for client_id, model in enumerate(
    personalized_v2_models
):

    probabilities = model.predict(
        X_test,
        verbose=0
    )

    predictions = (
        probabilities >= 0.5
    ).astype(int).flatten()

    acc = accuracy_score(
        y_test,
        predictions
    )

    prec = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    rec = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1_value = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    personalized_v2_results.append({
        "Client": client_id + 1,
        "Accuracy": acc * 100,
        "Precision": prec * 100,
        "Recall": rec * 100,
        "F1": f1_value * 100
    })

    print(
        f"Client {client_id + 1} | "
        f"Accuracy: {acc * 100:.2f}% | "
        f"Precision: {prec * 100:.2f}% | "
        f"Recall: {rec * 100:.2f}% | "
        f"F1: {f1_value * 100:.2f}%"
    )
```

```python
# ============================================
# STEP 23: CLIENT-LEVEL TRAIN/TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split

personalized_clients = []

print("=" * 70)
print("CLIENT-LEVEL TRAIN/TEST SPLIT")
print("=" * 70)

for client_id, client in enumerate(
    client_data_controlled
):

    X_client = client["X"]
    y_client = client["y"]

    X_local_train, X_local_test, y_local_train, y_local_test = (
        train_test_split(
            X_client,
            y_client,
            test_size=0.20,
            random_state=42,
            stratify=y_client
        )
    )

    personalized_clients.append({
        "X_train": X_local_train,
        "y_train": y_local_train,
        "X_test": X_local_test,
        "y_test": y_local_test
    })

    print(f"\nClient {client_id + 1}")

    print(
        "Training samples:",
        len(y_local_train)
    )

    print(
        "Testing samples :",
        len(y_local_test)
    )

    print(
        "Training classes:",
        np.bincount(y_local_train)
    )

    print(
        "Testing classes :",
        np.bincount(y_local_test)
    )

print("\n" + "=" * 70)
print("CLIENT SPLIT COMPLETE")
print("=" * 70)
```

```python
# ============================================
# STEP 24A: CLIENT-SPECIFIC PERSONALIZATION
# ============================================

from tensorflow.keras.callbacks import EarlyStopping

client_personalized_models = []

print("=" * 70)
print("CLIENT-SPECIFIC PERSONALIZATION")
print("=" * 70)

for client_id, client in enumerate(
    personalized_clients
):

    X_local_train = client["X_train"]
    y_local_train = client["y_train"]

    # Create copy of global FedAvg model
    personalized_model = create_model()

    personalized_model.set_weights(
        best_controlled_fedavg.get_weights()
    )

    # Freeze first shared layer
    personalized_model.layers[0].trainable = False

    # Adapt higher-level representation + classifier
    personalized_model.layers[1].trainable = True
    personalized_model.layers[2].trainable = True

    personalized_model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.0001
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    early_stopping = EarlyStopping(
        monitor="loss",
        patience=2,
        restore_best_weights=True
    )

    personalized_model.fit(
        X_local_train,
        y_local_train,
        epochs=20,
        batch_size=8,
        verbose=0,
        callbacks=[early_stopping]
    )

    client_personalized_models.append(
        personalized_model
    )

    print(
        f"Client {client_id + 1} | "
        f"Training samples: {len(y_local_train)} | "
        f"Personalization complete"
    )
```
